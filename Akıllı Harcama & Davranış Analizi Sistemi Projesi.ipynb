```python
cursor.execute("SELECT user_ID, SUM(amount) FROM Expenses GROUP BY user_ID ORDER BY SUM(amount) DESC")
kullanici_toplamlari = cursor.fetchall()
for toplam in kullanici_toplamlari:
    print(toplam)
```

```python
kullanicilar = {1:"Esmanur" , 2:"Edanur"}
butce_siniri = 700.0
for toplam in kullanici_toplamlari:
    kullanici_ID = toplam[0]
    tutar = toplam[1]
    isim = kullanicilar[kullanici_ID]
    print(f"{isim} adlı kullanıcının toplam harcaması {tutar} TL'dir.")
    print("-"* 40)
    if (tutar > 700):  
        print(f"⚠️ Dikkat {isim}, {butce_siniri} TL olan bütçeyi aştın!")
        print("-"* 40)
    else: 
        print(f"✅ Tebrikler {isim}, bütçeye uygun harcadın.")
```

```python
genel_toplam = 0
for toplam in kullanici_toplamlari:
    genel_toplam += toplam[1]
print(f"Evdeki Genel Toplam Harcama: {genel_toplam} TL")
print("-" * 40)
    
for toplam in kullanici_toplamlari:
    kullanici_ID = toplam[0]
    tutar = toplam[1]
    isim = kullanicilar[kullanici_ID]
    
    yuzde = (tutar /genel_toplam ) * 100 
    yuzde_yuvarlanmis = round(yuzde, 1)
    print(f"{isim}, evdeki toplam harcamaların %{yuzde_yuvarlanmis} 'ini gerçekleştirdi.")
```
